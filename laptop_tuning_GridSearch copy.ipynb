```python
import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV,RandomizedSearchCV
from sklearn.metrics import accuracy_score,r2_score
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import joblib
import pickle
```

```python
# SUPERVIZED LEARNING  GRIDSEARCHCV
```

```python
df=pd.read_csv("C:/Users/amank/Downloads/Learning_day_1/laptop_data.csv")
df.head(5)
```

```python
df.drop(columns=['Company','TypeName','TouchScreen'],inplace=True)
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Cpu_brand']=df['Cpu_brand'].fillna(df['Cpu_brand'].mode()[0])
df['Gpu_brand']=df['Gpu_brand'].fillna(df['Gpu_brand'].mode()[0])
df['Os']=df['Os'].fillna(df['Os'].mode()[0])
```

```python
df.head()
```

```python
label=LabelEncoder()
```

```python
df['encoded_Cpu_brand']=label.fit_transform(df['Cpu_brand'])
df['encoded_Gpu_brand']=label.fit_transform(df['Gpu_brand'])
df['encoded_Os']=label.fit_transform(df['Os'])
```

```python
df.head()
```

```python
df.drop(columns=['Cpu_brand','Gpu_brand','Os'],inplace=True)
```

```python
df.head()
```

```python
x=df[['Ram','Weight','Price','Ips','Ppi','HDD','SSD','encoded_Cpu_brand','encoded_Gpu_brand','encoded_Os']] 
y=df['Price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
best_param={'n_estimators':[100,200,300,400,500],'max_depth':[None,5,10,15,20]}
```

```python
model=RandomForestRegressor(random_state=42)
```

```python
grid_search=GridSearchCV(estimator=model,
                         
                         param_grid=best_param,

                         cv=5,

                         n_jobs=-1,
                         
                         verbose=2)
```

```python
grid_search.fit(x_train,y_train)
```

```python
best_model=grid_search.best_estimator_
```

```python
y_pred=best_model.predict(x_train)
r2_score_train=r2_score(y_train,y_pred)
r2_score_train
```

```python
y_pred=best_model.predict(x_test)
r2_score_test=r2_score(y_test,y_pred)
r2_score_test
```

```python
joblib.dump(best_model,'laptop_price_model.pkl')
```
